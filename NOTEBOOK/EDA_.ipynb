## Feature Engineering

```python
df['churn_flag'] = df['churn'].map({'Yes': 1, 'No': 0})
```

```python
df['customer_lifetime_value'] = df['monthly_charges'] * df['tenure']
```

```python
df['revenue_at_risk'] = np.where(
    df['churn_flag'] == 1,
    df['monthly_charges'],
    0
)
```

```python
df['tenure_bucket'] = pd.cut(
    df['tenure'],
    bins=[0, 6, 12, 24, np.inf],
    labels=['0-6 months', '6-12 months', '12-24 months', '24+ months'],
    include_lowest=True
)
```

```python
contract_risk_map = {
    'Month-to-month': 'High',
    'One year': 'Medium',
    'Two year': 'Low'
}

df['contract_risk'] = df['contract'].map(contract_risk_map)
```

```python
high_charge_threshold = df['monthly_charges'].quantile(0.75)

df['price_sensitivity_flag'] = np.where(
    (df['monthly_charges'] >= high_charge_threshold) &
    (df['tenure'] < 12),
    1,
    0
)
```

```python
feature_cols = [
    'customer_lifetime_value',
    'revenue_at_risk',
    'tenure_bucket',
    'contract_risk',
    'price_sensitivity_flag'
]

df[feature_cols].head()
```

### Tenure Cohort Summary: Customer Distribution, Churn, and Revenue Impact

```python
tenure_cohort_summary = (
    df.groupby('tenure_bucket', observed=True)
      .agg(
          customer_share=('customer_id', 'count'),
          revenue_at_risk=('revenue_at_risk', 'sum'),
          churn_rate=('churn_flag', 'mean')
      )
)

# Convert counts to percentage
tenure_cohort_summary['customer_share_pct'] = (
    tenure_cohort_summary['customer_share'] /
    tenure_cohort_summary['customer_share'].sum()
)

# Sort by revenue impact
tenure_cohort_summary = tenure_cohort_summary.sort_values(
    by='revenue_at_risk',
    ascending=False
)

tenure_cohort_summary
```

**Insight:**  
Although customers in their first 6 months represent only ~21% of the customer base, they account for the highest churn rate (~53%) and the largest share of revenue at risk, making early-stage retention the most critical business priority.

### Contract Risk Summary: Customer Distribution, Churn, and Revenue Impact

```python
contract_risk_summary = (
    df.groupby('contract_risk', observed=True)
      .agg(
          customer_count=('customer_id', 'count'),
          churn_rate=('churn_flag', 'mean'),
          revenue_at_risk=('revenue_at_risk', 'sum')
      )
)

# Convert count to percentage
contract_risk_summary['customer_share_pct'] = (
    contract_risk_summary['customer_count'] /
    contract_risk_summary['customer_count'].sum()
)

# Sort by churn severity
contract_risk_summary = contract_risk_summary.sort_values(
    by='churn_rate',
    ascending=False
)

contract_risk_summary
```

**Insight:**  
Month-to-month customers represent over half of the customer base and exhibit a significantly higher churn rate (~43%), contributing to the majority of revenue at risk and highlighting contract type as a critical retention lever.
